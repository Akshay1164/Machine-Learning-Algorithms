Random Forest 🌲🌲🌲

Now comes the interesting part.

Instead of using one Decision Tree, what if we create:

Tree 1
Tree 2
Tree 3
Tree 4
Tree 5
...
Tree 100

Each tree gives a prediction.

Then we take a vote.

For example:

Tree 1 → Approve
Tree 2 → Approve
Tree 3 → Reject
Tree 4 → Approve
Tree 5 → Approve

Vote:

Approve = 4
Reject  = 1

Final prediction:

Approve

That's a Random Forest.

7. Why is it called Random Forest?

Because we're creating a forest of many decision trees.

                 Random Forest
                      |
        ┌─────────────┼─────────────┐
        ↓             ↓             ↓
      Tree 1        Tree 2        Tree 3
        ↓             ↓             ↓
     Approve        Reject        Approve
        \             |             /
         \            |            /
          └───────────┼───────────┘
                      ↓
                   APPROVE

The important idea:

Many weak/independent trees together can produce a strong model.

```python
from sklearn.ensemble import RandomForestClassifier

X = [
    [22, 25, 550],
    [25, 30, 580],
    [28, 40, 620],
    [30, 50, 680],
    [35, 60, 700],
    [40, 70, 720],
    [45, 80, 750],
    [50, 90, 780]
]

y = [0, 0, 0, 1, 1, 1, 1, 1]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

n_estimators=100

customer = [[32, 55, 690]]

prediction = model.predict(customer)

print("Prediction:", prediction[0])
```
