Now XGBoost ⭐

This is where things become particularly useful for real-world ML.

XGBoost = Extreme Gradient Boosting.

It's an optimized and highly effective implementation of gradient-boosted decision trees.

It became extremely popular because it performs very well on structured/tabular data.

You will see XGBoost frequently in:

Kaggle competitions
Credit risk
Fraud detection
Customer churn
Recommendation systems
Business prediction problems

```python
from xgboost import XGBClassifier

X = [
    [22, 25, 550],
    [25, 30, 580],
    [28, 40, 620],
    [30, 50, 680],
    [35, 60, 700],
    [40, 70, 720],
    [45, 80, 750],
    [50, 90, 780]
]

y = [0, 0, 0, 1, 1, 1, 1, 1]

model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

model.fit(X, y)

customer = [[32, 55, 690]]

prediction = model.predict(customer)

print("Prediction:", prediction[0])
```

So what's the difference?

This is the most important section.

Algorithm	Basic Idea
Decision Tree	One tree makes decisions
Random Forest	Many trees vote
Gradient Boosting	Trees sequentially correct errors
XGBoost	Optimized/advanced gradient boosting

Think of four people:

Decision Tree
👨
One expert makes the decision.
Random Forest
👨 👩 👨 👩 👨
Many experts vote.
Gradient Boosting
👨 → 👩 → 👨 → 👩
Each expert learns from previous mistakes.
XGBoost
🚀
A highly optimized and powerful version
of gradient-boosted trees.
